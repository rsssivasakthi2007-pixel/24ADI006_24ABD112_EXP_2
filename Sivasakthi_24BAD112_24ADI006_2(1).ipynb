```python
print("Roll No: 24BAD112")
print("Name: SIVASAKTHI S")

import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(-10, 10, 400)

# Activation functions
def sigmoid(x):
    return 1 / (1 + np.exp(-x))
def tanh(x):
    return np.tanh(x)
def relu(x):
    return np.maximum(0, x)

# Derivatives (for gradient behavior)
def sigmoid_derivative(x):
    s = sigmoid(x)
    return s * (1 - s)
def tanh_derivative(x):
    return 1 - np.tanh(x)**2
def relu_derivative(x):
    return np.where(x > 0, 1, 0)

# Plot activation functions
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.plot(x, sigmoid(x), label='Sigmoid', color='blue')
plt.title('Sigmoid Activation Function')
plt.xlabel('x'); plt.ylabel('f(x)')
plt.grid(True); plt.legend()

plt.subplot(1, 3, 2)
plt.plot(x, tanh(x), label='Tanh', color='green')
plt.title('Tanh Activation Function')
plt.xlabel('x'); plt.ylabel('f(x)')
plt.grid(True); plt.legend()

plt.subplot(1, 3, 3)
plt.plot(x, relu(x), label='ReLU', color='red')
plt.title('ReLU Activation Function')
plt.xlabel('x'); plt.ylabel('f(x)')
plt.grid(True); plt.legend()

plt.tight_layout()
plt.show()

# Plot derivatives (gradient behavior)
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.plot(x, sigmoid_derivative(x), label="Sigmoid'", color='blue')
plt.title('Sigmoid Derivative (Gradient)')
plt.xlabel('x'); plt.ylabel("f'(x)")
plt.grid(True); plt.legend()

plt.subplot(1, 3, 2)
plt.plot(x, tanh_derivative(x), label="Tanh'", color='green')
plt.title('Tanh Derivative (Gradient)')
plt.xlabel('x'); plt.ylabel("f'(x)")
plt.grid(True); plt.legend()

plt.subplot(1, 3, 3)
plt.plot(x, relu_derivative(x), label="ReLU'", color='red')
plt.title('ReLU Derivative (Gradient)')
plt.xlabel('x'); plt.ylabel("f'(x)")
plt.grid(True); plt.legend()

plt.tight_layout()
plt.show()

# Comparison summary table
print("Comparison of Activation Functions")
print("-" * 70)
print(f"{'Property':<25}{'Sigmoid':<20}{'Tanh':<15}{'ReLU'}")
print("-" * 70)
print(f"{'Output Range':<25}{'(0, 1)':<20}{'(-1, 1)':<15}{'[0, inf)'}")
print(f"{'Saturation':<25}{'Both ends':<20}{'Both ends':<15}{'Only negative side'}")
print(f"{'Max Gradient':<25}{'0.25':<20}{'1.0':<15}{'1 (for x>0)'}")
print(f"{'Computational Cost':<25}{'High (exp)':<20}{'High (exp)':<15}{'Low (simple max)'}")
```
